## 4. Validasi submission

Submission harus mempunyai jumlah baris yang sama dengan sample submission, ID yang sama, dan nilai target 1 sampai 5.

```python
if SUBMISSION_PATH.exists() and SUBMISSION_PATH.stat().st_size > 0:
    submission = pd.read_csv(SUBMISSION_PATH)
    sample = pd.read_csv(ROOT_DIR / 'data' / 'sample_submission.csv')
    allowed_labels = {1, 2, 3, 4, 5}

    assert list(submission.columns) == list(sample.columns)
    assert len(submission) == len(sample)
    assert submission['building_id'].equals(sample['building_id'])
    assert set(submission['damage_grade']).issubset(allowed_labels)
    assert pd.api.types.is_integer_dtype(submission['damage_grade'])
    assert not submission.isna().any().any()

    print('Submission valid:', submission.shape)
    display(submission.head())
    display(submission['damage_grade'].value_counts().sort_index())
else:
    print('submission.csv belum tersedia.')
```
